This notebook implements supervised machine learning using Python and Scikit-learn. It covers regression and classification, model training, prediction, evaluation, and model comparison.

## 1. Objective

The objective is to understand and implement supervised machine learning algorithms using Python and Scikit-learn. The workflow includes train-test splitting, model training, prediction, evaluation, and comparison of multiple models.

## 2. Tools and Technologies

- Python 3
- Scikit-learn
- Pandas
- NumPy
- Jupyter Notebook / Google Colab

````python
# Import libraries
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
from sklearn.datasets import load_diabetes, load_breast_cancer
````

## 3. Datasets

Two datasets available through Scikit-learn are used:

1. **Diabetes dataset** – used for regression. The target is a quantitative disease-progression measure.
2. **Breast Cancer Wisconsin dataset** – used for binary classification.

Both datasets are provided by Scikit-learn, so no external dataset download is required.

````python
# Load the Diabetes dataset for regression
diabetes = load_diabetes()
X_reg = diabetes.data
y_reg = diabetes.target

print("Diabetes dataset shape:", X_reg.shape)
print("Regression target shape:", y_reg.shape)
````

````python
# Load the Breast Cancer dataset for classification
breast_cancer = load_breast_cancer()
X_cls = breast_cancer.data
y_cls = breast_cancer.target

print("Breast Cancer dataset shape:", X_cls.shape)
print("Classification target shape:", y_cls.shape)
````

## 4. Regression – Train/Test Split

The regression data is divided into training and testing sets using an **80:20 split** with `random_state=42`.

````python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.20, random_state=42
)

print("Training samples:", X_train_reg.shape[0])
print("Testing samples:", X_test_reg.shape[0])
````

## 5. Linear Regression

Linear Regression models the relationship between input features and a continuous target using a linear function.

````python
linear_model = LinearRegression()
linear_model.fit(X_train_reg, y_train_reg)

linear_predictions = linear_model.predict(X_test_reg)

linear_mae = mean_absolute_error(y_test_reg, linear_predictions)
linear_rmse = np.sqrt(mean_squared_error(y_test_reg, linear_predictions))
linear_r2 = r2_score(y_test_reg, linear_predictions)

print("Linear Regression")
print("MAE:", round(linear_mae, 3))
print("RMSE:", round(linear_rmse, 3))
print("R²:", round(linear_r2, 3))
````

## 6. Random Forest Regression

Random Forest Regressor combines multiple decision trees and can capture more complex relationships.

````python
rf_model = RandomForestRegressor(random_state=42)
rf_model.fit(X_train_reg, y_train_reg)

rf_predictions = rf_model.predict(X_test_reg)

rf_mae = mean_absolute_error(y_test_reg, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_test_reg, rf_predictions))
rf_r2 = r2_score(y_test_reg, rf_predictions)

print("Random Forest Regressor")
print("MAE:", round(rf_mae, 3))
print("RMSE:", round(rf_rmse, 3))
print("R²:", round(rf_r2, 3))
````

## 7. Regression Model Comparison

Regression models are evaluated using **MAE, RMSE, and R²** because the target variable is continuous.

````python
regression_results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest Regressor"],
    "MAE": [linear_mae, rf_mae],
    "RMSE": [linear_rmse, rf_rmse],
    "R²": [linear_r2, rf_r2]
})

regression_results.round(3)
````

### Regression Metrics

- **MAE (Mean Absolute Error):** average absolute difference between predicted and actual values.
- **RMSE (Root Mean Squared Error):** gives greater weight to larger errors.
- **R²:** indicates the proportion of target variation explained by the model.

## 8. Classification – Train/Test Split

The Breast Cancer dataset is split using an **80:20 stratified split** so the class distribution is maintained between training and testing data.

````python
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls, y_cls, test_size=0.20, random_state=42, stratify=y_cls
)

print("Training samples:", X_train_cls.shape[0])
print("Testing samples:", X_test_cls.shape[0])
````

## 9. Logistic Regression

Logistic Regression estimates class probabilities using a logistic function.

````python
logistic_model = LogisticRegression(max_iter=5000)
logistic_model.fit(X_train_cls, y_train_cls)

logistic_predictions = logistic_model.predict(X_test_cls)

logistic_accuracy = accuracy_score(y_test_cls, logistic_predictions)
logistic_precision = precision_score(y_test_cls, logistic_predictions)
logistic_recall = recall_score(y_test_cls, logistic_predictions)
logistic_f1 = f1_score(y_test_cls, logistic_predictions)

print("Logistic Regression")
print("Accuracy:", round(logistic_accuracy, 3))
print("Precision:", round(logistic_precision, 3))
print("Recall:", round(logistic_recall, 3))
print("F1 Score:", round(logistic_f1, 3))
````

## 10. Decision Tree Classifier

A Decision Tree creates a sequence of feature-based decisions to classify observations.

````python
tree_model = DecisionTreeClassifier(random_state=42)
tree_model.fit(X_train_cls, y_train_cls)

tree_predictions = tree_model.predict(X_test_cls)

tree_accuracy = accuracy_score(y_test_cls, tree_predictions)
tree_precision = precision_score(y_test_cls, tree_predictions)
tree_recall = recall_score(y_test_cls, tree_predictions)
tree_f1 = f1_score(y_test_cls, tree_predictions)

print("Decision Tree Classifier")
print("Accuracy:", round(tree_accuracy, 3))
print("Precision:", round(tree_precision, 3))
print("Recall:", round(tree_recall, 3))
print("F1 Score:", round(tree_f1, 3))
````

## 11. Classification Model Comparison

Classification models are evaluated using **Accuracy, Precision, Recall, and F1 Score**.

````python
classification_results = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree Classifier"],
    "Accuracy": [logistic_accuracy, tree_accuracy],
    "Precision": [logistic_precision, tree_precision],
    "Recall": [logistic_recall, tree_recall],
    "F1 Score": [logistic_f1, tree_f1]
})

classification_results.round(3)
````

### Classification Metrics

- **Accuracy:** proportion of correctly classified observations.
- **Precision:** how many predicted positive cases were actually positive.
- **Recall:** how many actual positive cases were identified.
- **F1 Score:** combined measure of precision and recall.

## 12. Overall Observations

Different supervised learning algorithms can produce different results on the same problem. For regression, MAE, RMSE and R² are appropriate because the target is continuous. For classification, accuracy, precision, recall and F1 score are used. The comparison demonstrates the importance of selecting evaluation metrics that match the machine learning problem and practical objective.
